## Import Libs

```python
import pandas as pd
from sklearn.preprocessing import LabelEncoder, StandardScaler
import ast
from sklearn.preprocessing import MultiLabelBinarizer
import re
import numpy as np
pd.set_option('display.max_columns', None)
pd.set_option('display.max_rows', None)
pd.set_option('display.width', None)
pd.set_option('display.max_colwidth', None)
```

## Load Data

```python
def load_data(path):
    try:
        df = pd.read_csv(path)
        print(f'Data is loaded with : {len(df)} Rows')
        return df
    except Exception as e:
        print(f'You have a problem: {e}')
        return None

path = "../../data/raw/Clean_1.csv"
df = load_data(path)
```

## Encoded Features

### Job Titel Encode 

```python
le_job = LabelEncoder()
df['job_clean_encoded'] = le_job.fit_transform(df['job_clean'])
print(f"job_clean classes: {le_job.classes_}")
```

### Label encode Size 

```python
# Label encode Size
size_order = {
    '1 to 50 employees': 1,
    '51 to 200 employees': 2,
    '201 to 500 employees': 3,
    '501 to 1000 employees': 4,
    '1001 to 5000 employees': 5,
    '5001 to 10000 employees': 6,
    '10000+ employees': 7,
    'Unknown': 0
}
df['size_encoded'] = df['Size'].map(size_order).fillna(0)
df['size_encoded'].value_counts()
```

### Label encode Sector

```python
# Label encode Sector
le_sector = LabelEncoder()
df['sector_encoded'] = le_sector.fit_transform(df['Sector'].astype(str))
print(f"Unique sectors: {len(le_sector.classes_)}")
print(f"Unique sectors: {le_sector.classes_}")
df['sector_encoded'].value_counts()
```

### Label encode Industry

```python
# Label encode Industry
le_industry = LabelEncoder()
df['industry_encoded'] = le_industry.fit_transform(df['Industry'].astype(str))
print(f"Unique industries: {len(le_industry.classes_)}")
print(f"Unique industries: {le_industry.classes_}")
```

### State encoding

```python
# State encoding
le_state = LabelEncoder()
df['state_encoded'] = le_state.fit_transform(df['state'].fillna('Unknown'))
print(f"Unique states: {len(le_state.classes_)}")
print(f"Unique states: {le_state.classes_}")
df['state_encoded'].value_counts()
```

```python
# Handle Target
# Ensure salary_clean is numeric and no NaNs
df = df.dropna(subset=['salary_clean'])
```

```python
# Drop original columns that are no longer needed
cols_to_drop = [
    "index", "Job Title", "Salary Estimate", "Job Description", "Location", 
    "Company Name", "Headquarters", "Size", "Founded", "Job_clean", "Type of ownership", 
    "Industry", "Sector", "Revenue", "Competitors", "description_clean", "job_clean", "state",
    "skills", "skills_list"
]
df_final = df.drop(columns=[col for col in cols_to_drop if col in df.columns])
    
# Fill remaining NaNs (e.g. Rating, company_age) if any
df_final.fillna(df_final.median(), inplace=True)
    
print(f"Preprocessed data shape: {df_final.shape}")
```
