## Import Libs

```python
import pandas as pd
from sklearn.preprocessing import LabelEncoder, StandardScaler
from sklearn.feature_extraction.text import TfidfVectorizer
```

## Load Data

```python
def load_data(path):
    try:
        df = pd.read_csv(path)
        print(f'Data is loaded with : {len(df)} Rows')
        return df
    except Exception as e:
        print(f'You have a problem: {e}')
        return None

path = "../../data/raw/Clean_1.csv"
df = load_data(path)
df.head()
```

## Encoded Features

### Job Titel Encode 

```python
le_job = LabelEncoder()
df['job_clean_encoded'] = le_job.fit_transform(df['job_clean'])
print(f"job_clean classes: {le_job.classes_}")
```

### Label encode Size 

```python
# Label encode Size
size_order = {
    '1 to 50 employees': 1,
    '51 to 200 employees': 2,
    '201 to 500 employees': 3,
    '501 to 1000 employees': 4,
    '1001 to 5000 employees': 5,
    '5001 to 10000 employees': 6,
    '10000+ employees': 7,
    'Unknown': 0
}
df['size_encoded'] = df['Size'].map(size_order).fillna(0)
df['size_encoded'].value_counts()
```

### Label encode Sector

```python
# Label encode Sector
le_sector = LabelEncoder()
df['sector_encoded'] = le_sector.fit_transform(df['Sector'].astype(str))
print(f"Unique sectors: {len(le_sector.classes_)}")
print(f"Unique sectors: {le_sector.classes_}")
df['sector_encoded'].value_counts()
```

### Label encode Industry

```python
# Label encode Industry
le_industry = LabelEncoder()
df['industry_encoded'] = le_industry.fit_transform(df['Industry'].astype(str))
print(f"Unique industries: {len(le_industry.classes_)}")
print(f"Unique industries: {le_industry.classes_}")
```

### State encoding

```python
# State encoding
le_state = LabelEncoder()
df['state_encoded'] = le_state.fit_transform(df['state'].fillna('Unknown'))
print(f"Unique states: {len(le_state.classes_)}")
print(f"Unique states: {le_state.classes_}")
df['state_encoded'].value_counts()
```

## Prepare Data for Model

###  Convert the Job Description text into a numerical matrix 

```python
from numpy.ma import max
from sklearn.feature_extraction.text import TfidfVectorizer

vectorizing = TfidfVectorizer(max_features=5000)
X_txt = vectorizing.fit_transform(df['Job Description'])
X_txt.shape
```

### Git other Features 

```python
X_numeric = df[["company_age", "skills_count", "desc_length", "title_length", "job_clean_encoded", "size_encoded", "sector_encoded", "industry_encoded", "state_encoded"]]
X_numeric
```

### Apply Standardscaler for numircal

```python
from sklearn.preprocessing import StandardScaler

# Numeric features
numeric_cols = ["company_age", "skills_count", "desc_length", "title_length"]
scaler = StandardScaler()
X_numeric_scaled = scaler.fit_transform(df[numeric_cols])
X_numeric_scaled
```

```python
categorical_cols = ["size_encoded", "sector_encoded", "industry_encoded", "state_encoded"]
X_categorical = df[categorical_cols].values

X_categorical
```

### Merge :

```python
from scipy.sparse import hstack

X = hstack([X_txt, X_numeric_scaled, X_categorical])
X.shape
```

### Target

```python
y = df["salary_clean"]  
y
```

## Save Last Data

```python
import numpy as np
from scipy import sparse

# Save sparse X
sparse.save_npz("../../data/raw/X_tfidf_features.npz", X)

# Save target
np.save("../../data/raw/y_salary.npy", y)
```
